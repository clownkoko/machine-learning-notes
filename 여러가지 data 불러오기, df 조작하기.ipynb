```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn import datasets
```

# 꽃 구분하기
아이리스(붓꽃) 이라는 꽃에 관련된 데이터입니다.
아이리스의 <span style="color:red">**피처 특성**</span>에 따라 <span style="color:red">**꽃의 종류**</span>를 분류하는 머신러닝 알고리즘을 만들 수 있습니다.


```python
# 아이리스 데이터 불러오기

raw_iris = datasets.load_iris()

#피처데이터, 타깃데이터 데이터 프레임 생성

x_iris = pd.DataFrame(raw_iris.data)
y_iris = pd.DataFrame(raw_iris.target)

#concat이용해 두 데이터 프레임 합치기

df_iris = pd.concat([x_iris, y_iris], axis = 1)

df_iris.head()
```

`feature_names` 를 이용해 피처 이름 출력합니다.
아이리스의 데이터 피처는 총 4개로 구성됨을 알 수 있습니다.

```python
feature_iris = raw_iris.feature_names
print(feature_iris)
```

```python
# 피처와 타깃 변수 이름을 append 이용해 합침

col_iris = np.append(feature_iris, ['target'])

#아이리스 전체 데이터의 열 이름 설정

df_iris.columns = col_iris

df_iris.head()
```
